## 1 · Anatomy of an LLM web API call
*No SDK, no magic: one HTTPS POST with a JSON body*

Every LLM API call has the same four parts:

1. **Endpoint** — a URL, here `POST /v1/responses`
2. **Authentication** — an `Authorization: Bearer <key>` header
3. **Request body** — JSON with the `model` and your `input`
4. **Response body** — JSON with the generated `output` and the `usage` (tokens = what you pay for)

We use plain `requests` first so nothing is hidden.

````python
payload = {
    "model": OPENAI_MODEL,
    "input": "In one sentence: what is a web API?",
    "reasoning": {"effort": "minimal"},   # gpt-5 models "think" before answering; keep that short and cheap
}

r = requests.post(
    f"{OPENAI_URL}/responses",
    headers={"Authorization": f"Bearer {OPENAI_API_KEY}", "Content-Type": "application/json"},
    json=payload,
    timeout=60,
)

print("HTTP status:", r.status_code)
data = r.json()
print(json.dumps(data, indent=2)[:1800])
````

````python
# The answer is nested inside the JSON: output -> message -> content -> output_text
answer = "".join(
    part["text"]
    for item in data["output"] if item["type"] == "message"
    for part in item["content"] if part["type"] == "output_text"
)
print("ANSWER:", answer)
print("USAGE: ", data["usage"])
````

## 3 · NVIDIA Nemotron via Hugging Face
*Same SDK, different `base_url`: switching providers is a one-line change*

NVIDIA publishes the **Nemotron** family as open weights on the Hugging Face Hub. You could download them and run them on your own GPU; today we call them remotely through Hugging Face **Inference Providers**, a router that forwards your request to a company hosting the model.

That router speaks the *Chat Completions* protocol, the de-facto standard that almost every LLM provider implements. So we reuse the `openai` package and only change `base_url`, the key and the model name.

| | OpenAI | NVIDIA model on Hugging Face |
|---|---|---|
| Base URL | `api.openai.com/v1` | `router.huggingface.co/v1` |
| Key | `OPENAI_API_KEY` | `HF_KEY` |
| Endpoint | `/responses` | `/chat/completions` |
| Weights | closed | open (you can self-host) |

````python
def clean(text):
    # Some open models print their reasoning inside <think>...</think>; keep only the final answer.
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.S).strip()

def nvidia_chat(messages, max_retries=5, backoff_factor=2, **params):
    global NVIDIA_MODEL, HF_KEY
    # Lista de IPs públicos alternativos do Hugging Face para contornar falhas de DNS
    fallback_ips = ["18.215.226.54", "54.235.119.141", "3.218.172.193"]
    url_dns = "https://api-inference.huggingface.co/v1/chat/completions"

    headers = {
        "Authorization": f"Bearer {HF_KEY}",
        "Content-Type": "application/json",
        "Host": "api-inference.huggingface.co"
    }
    payload = {
        "model": NVIDIA_MODEL,
        "messages": messages,
        **params
    }

    last_exception = None
    for attempt in range(max_retries):
        # Se for uma tentativa avançada, reveza entre DNS normal e IPs diretos
        use_ip_fallback = (attempt >= 1)
        if use_ip_fallback:
            ip = fallback_ips[attempt % len(fallback_ips)]
            url = f"https://{ip}/v1/chat/completions"
        else:
            url = url_dns

        try:
            # verify=False pode ser necessário ao acessar via IP direto para evitar erros de certificado SSL (SNI)
            response = requests.post(url, headers=headers, json=payload, timeout=20, verify=not use_ip_fallback)
            if response.status_code == 200:
                data = response.json()

                class SimpleUsage:
                    def __init__(self, usage_dict):
                        self.prompt_tokens = usage_dict.get("prompt_tokens", 0)
                        self.completion_tokens = usage_dict.get("completion_tokens", 0)

                class SimpleResponse:
                    def __init__(self, data):
                        self.model = data.get("model", NVIDIA_MODEL)
                        self.usage = SimpleUsage(data.get("usage", {}))

                raw_response = SimpleResponse(data)
                return clean(data["choices"][0]["message"]["content"]), raw_response
            else:
                raise RuntimeError(f"Erro {response.status_code}: {response.text}")
        except Exception as e:
            last_exception = e
            if attempt < max_retries - 1:
                sleep_time = backoff_factor ** (attempt + 1)
                mode = f"usando IP estático {fallback_ips[attempt % len(fallback_ips)]}" if use_ip_fallback else "usando DNS normal"
                print(f"[Falha {mode}. Tentando novamente em {sleep_time}s... Erro: {e}]")
                time.sleep(sleep_time)
            else:
                raise RuntimeError(f"Não foi possível conectar após {max_retries} tentativas. Erro original: {last_exception}")

try:
    # Desabilita temporariamente avisos de conexão insegura caso use IP de fallback sem verificação de certificado estrita
    import urllib3
    urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

    text, raw = nvidia_chat([
        {"role": "system", "content": "You are a teaching assistant for CS undergraduates. Be concise."},
        {"role": "user",   "content": "What does 'open weights' mean for an LLM, and why would a company care?"},
    ])
    display(Markdown(text))
    print(f"\nmodel: {raw.model} | prompt tokens: {raw.usage.prompt_tokens} | completion tokens: {raw.usage.completion_tokens}")
except Exception as e:
    print(f"Não foi possível conectar ao Hugging Face: {e}")
````

## 4 · Roles, messages and memory
*The API is stateless: the model remembers only what you send again*

A chat is a list of messages, each with a **role**:

- `system` — rules and persona, set by the developer
- `user` — what the human typed
- `assistant` — what the model answered before

The server keeps **nothing** between calls. "Memory" in a chatbot is your Python code resending the whole history every time — which is also why long chats cost more.

````python
# Duas chamadas independentes: a segunda chamada não tem ideia do que a primeira disse.
try:
    a1, _ = nvidia_chat([{"role": "user", "content": "My name is Ada and I study at PUC-SP. Say hi."}])
    a2, _ = nvidia_chat([{"role": "user", "content": "What is my name and where do I study?"}])
    print("call 1:", a1)
    print("call 2:", a2)
except Exception as e:
    print(f"Não foi possível conectar ao Hugging Face devido a um erro de rede: {e}")
````

````python
class Chat:
    # A minimal chatbot: the 'memory' is just a Python list that we resend on every call.
    def __init__(self, system):
        self.messages = [{"role": "system", "content": system}]

    def say(self, text):
        self.messages.append({"role": "user", "content": text})
        answer, raw = nvidia_chat(self.messages)
        self.messages.append({"role": "assistant", "content": answer})
        print(f"you > {text}\nbot > {answer}\n      [{raw.usage.prompt_tokens} prompt tokens sent]\n")

chat = Chat("You are a friendly assistant. Answer in one short sentence.")
chat.say("My name is Ada and I study at PUC-SP.")
chat.say("I am learning how LLM APIs work.")
chat.say("What is my name, where do I study, and what am I learning?")
````

Watch the *prompt tokens* grow on every turn: you pay again for the whole history.

## 5 · Streaming
*Tokens arrive one by one over an open HTTP connection (Server-Sent Events)*

Without streaming you wait for the full answer. With `stream=True` the server pushes small chunks as they are generated — this is what makes chat UIs feel fast. Two numbers define the user experience:

- **TTFT** (time to first token) — how long until *something* appears
- **Throughput** — how fast the text grows after that

We stream the same prompt from both providers and record a timestamp for every chunk.

````python
PROMPT = "Explain in about 120 words how HTTPS protects an API key in transit."

def stream_openai(prompt):
    t0, rows = time.perf_counter(), []
    stream = openai_client.responses.create(
        model=OPENAI_MODEL, input=prompt, reasoning={"effort": "minimal"}, stream=True
    )
    for event in stream:
        if event.type == "response.output_text.delta":      # many event types exist; we want the text deltas
            print(event.delta, end="", flush=True)
            rows.append((time.perf_counter() - t0, event.delta))
    return rows

def stream_nvidia(prompt):
    t0, rows = time.perf_counter(), []
    try:
        stream = nvidia_client.chat.completions.create(
            model=NVIDIA_MODEL, messages=[{"role": "user", "content": prompt}], stream=True
        )
        for chunk in stream:
            if chunk.choices and chunk.choices[0].delta.content:
                piece = chunk.choices[0].delta.content
                print(piece, end="", flush=True)
                rows.append((time.perf_counter() - t0, piece))
    except Exception as e:
        print(f"\n[Erro de conexão no streaming da NVIDIA: {e}]")
    return rows

print("── OpenAI ──")
rows_openai = stream_openai(PROMPT)
print("\n\n── NVIDIA ──")
rows_nvidia = stream_nvidia(PROMPT)
````

````python
def to_frame(rows, provider):
    df = pd.DataFrame(rows, columns=["seconds", "chunk"])
    df["chars"] = df["chunk"].str.len().cumsum()
    df["provider"] = provider
    return df

stream_df = pd.concat([to_frame(rows_openai, "OpenAI"), to_frame(rows_nvidia, "NVIDIA")], ignore_index=True)

fig, ax = plt.subplots(figsize=(8, 4))
sns.lineplot(data=stream_df, x="seconds", y="chars", hue="provider", drawstyle="steps-post", ax=ax)
ax.set(title="Streaming: characters received over time", xlabel="seconds since request", ylabel="characters received")
plt.show()

stream_df.groupby("provider").agg(ttft_s=("seconds", "min"), total_s=("seconds", "max"), chars=("chars", "max")).round(2)
````

## 6 · Temperature and sampling
*An LLM is a probability distribution; temperature decides how adventurous the dice are*

At each step the model produces a probability for every possible next token. **Temperature** rescales those probabilities before one token is drawn:

- `0` → always pick the most likely token: repeatable, good for code, extraction, grading
- `~0.7` → balanced default
- `>1` → flatter distribution: more surprising, more errors

Experiment: the same creative prompt, 6 times at each of 3 temperatures. We count how many *different* answers come back. (We use the NVIDIA model here: OpenAI's gpt-5 reasoning models do not accept a `temperature` parameter.)

````python
QUESTION = "Invent a name for a coffee shop run by robots. Reply with the name only, nothing else."
TEMPERATURES, RUNS = [0.0, 0.7, 1.4], 6

rows = []
for t in TEMPERATURES:
    for run in range(RUNS):
        try:
            answer, _ = nvidia_chat([{"role": "user", "content": QUESTION}], temperature=t, max_tokens=200)
            clean_ans = answer.splitlines()[-1].strip(' "*.') if answer else ""
            rows.append({"temperature": t, "run": run + 1, "answer": clean_ans})
        except Exception as e:
            print(f"[Erro na temp {t}, run {run+1}]: {e}")
            rows.append({"temperature": t, "run": run + 1, "answer": "Erro de Conexão"})

temp_df = pd.DataFrame(rows)
temp_df.pivot(index="run", columns="temperature", values="answer")
````

````python
diversity = temp_df.groupby("temperature")["answer"].nunique().reset_index(name="unique_answers")

fig, ax = plt.subplots(figsize=(6, 3.5))
sns.barplot(data=diversity, x="temperature", y="unique_answers", color=sns.color_palette()[0], ax=ax)
ax.bar_label(ax.containers[0])
ax.set(title=f"Distinct answers out of {RUNS} runs", xlabel="temperature", ylabel="unique answers", ylim=(0, RUNS + 0.5))
plt.show()
````

## 7 · Structured outputs → pandas
*Make the model fill in a schema, so your program gets data instead of prose*

Free text is for humans. Programs need **typed fields**. With structured outputs you hand the API a JSON Schema (here generated from a Pydantic class) and the model is constrained to return JSON that matches it — no regex, no "please answer in JSON".

Use case: turn raw customer reviews (English and Portuguese mixed) into a table we can analyse.

````python
reviews = [
    "Arrived two days early and the packaging was perfect. Very happy!",
    "O produto quebrou na primeira semana. Péssima qualidade.",
    "Support took 40 minutes to answer and did not solve my problem.",
    "Preço justo, funciona como esperado.",
    "Too expensive for what it does. I regret buying it.",
    "A entrega atrasou dez dias e ninguém me avisou.",
    "Great build quality, feels premium. Worth every cent.",
    "O atendimento foi rápido e muito educado, resolveram tudo.",
    "It is okay. Nothing special, nothing wrong.",
    "Chegou no prazo, mas a caixa veio amassada.",
]

class Review(BaseModel):
    id: int
    language: Literal["en", "pt"]
    sentiment: Literal["positive", "neutral", "negative"]
    topic: Literal["delivery", "quality", "price", "support"]
    stars: int            # 1 (terrible) to 5 (excellent)
    summary_en: str       # at most 8 words, in English

class ReviewBatch(BaseModel):
    reviews: list[Review]

response = openai_client.responses.parse(
    model=OPENAI_MODEL,
    reasoning={"effort": "minimal"},
    instructions="Classify each customer review. Return one item per review, keeping its id.",
    input="\n".join(f"{i}: {text}" for i, text in enumerate(reviews)),
    text_format=ReviewBatch,          # <- the schema the answer must follow
)

reviews_df = pd.DataFrame([r.model_dump() for r in response.output_parsed.reviews]).set_index("id")
reviews_df
````

## 8 · Tool calling: the LLM queries SQL
*The model cannot run code; it asks **your** program to run it*

An LLM only produces text. **Tool calling** (function calling) is a protocol on top of that:

1. You describe your functions to the model (name, purpose, JSON parameters).
2. The model may answer with a *request* — "call `run_sql` with this query".
3. **Your code** executes it and sends the result back.
4. The model writes the final answer using that result.

This loop is the core of every "AI agent". Our tool: read-only SQL on a small grades database.

````python
db = sqlite3.connect(":memory:")
db.executescript('''
CREATE TABLE grades (student TEXT, course TEXT, semester TEXT, grade REAL);
INSERT INTO grades VALUES
 ('Ana','Algorithms','2026-1',8.5), ('Ana','Databases','2026-1',9.0), ('Ana','AI','2026-2',9.5),
 ('Bruno','Algorithms','2026-1',6.0), ('Bruno','Databases','2026-1',7.5), ('Bruno','AI','2026-2',5.5),
 ('Carla','Algorithms','2026-1',9.5), ('Carla','Databases','2026-1',8.0), ('Carla','AI','2026-2',10.0),
 ('Diego','Algorithms','2026-1',4.5), ('Diego','Databases','2026-1',6.5), ('Diego','AI','2026-2',7.0);
''')

def run_sql(query: str) -> str:
    # The tool. Only SELECT is allowed: never let a model write to your database.
    if not query.strip().lower().startswith("select"):
        return "ERROR: only SELECT statements are allowed."
    try:
        return pd.read_sql_query(query, db).to_json(orient="records")
    except Exception as e:
        return f"ERROR: {e}"

tools = [{
    "type": "function",
    "name": "run_sql",
    "description": "Run a read-only SQLite SELECT query on the university database and return the rows as JSON.",
    "parameters": {
        "type": "object",
        "properties": {"query": {"type": "string", "description": "A single SQLite SELECT statement."}},
        "required": ["query"],
        "additionalProperties": False,
    },
    "strict": True,
}]

SYSTEM = ("You answer questions about student grades. "
          "Database schema: grades(student TEXT, course TEXT, semester TEXT, grade REAL). "
          "Always use the run_sql tool to get data; never guess numbers.")

def ask_db(question, max_turns=5):
    trace = [("App", "LLM", question)]
    inputs = [{"role": "user", "content": question}]
    for _ in range(max_turns):
        response = openai_client.responses.create(
            model=OPENAI_MODEL, instructions=SYSTEM, input=inputs, tools=tools, reasoning={"effort": "low"}
        )
        inputs += response.output                                   # keep everything the model produced
        calls = [item for item in response.output if item.type == "function_call"]
        if not calls:                                               # no tool request -> this is the final answer
            trace.append(("LLM", "App", response.output_text))
            return response.output_text, trace
        for call in calls:
            query = json.loads(call.arguments)["query"]
            trace.append(("LLM", "App", f"call run_sql: {query}"))
            result = run_sql(query)                                 # <- OUR code runs the query, not the model
            trace += [("App", "SQLite", query), ("SQLite", "App", result), ("App", "LLM", f"tool result: {result}")]
            inputs.append({"type": "function_call_output", "call_id": call.call_id, "output": result})
    return "Stopped: too many turns.", trace

answer, trace = ask_db("Which student has the highest average grade, and what is that average?")
display(Markdown(answer))
````

````python
def draw_sequence(trace, actors=("App", "LLM", "SQLite"), width=46):
    # Draw the message trace as a sequence diagram: one vertical lifeline per actor, one arrow per message.
    x = {name: i for i, name in enumerate(actors)}
    fig, ax = plt.subplots(figsize=(10, 1.2 + 0.75 * len(trace)))
    for name, xi in x.items():
        ax.plot([xi, xi], [0.4, -len(trace) - 0.3], color="#bbbbbb", lw=1.5, zorder=1)
        ax.text(xi, 0.75, name, ha="center", va="center", fontsize=12, fontweight="bold",
                bbox=dict(boxstyle="round,pad=0.45", fc="#e8eef7", ec="#4c72b0"))
    for i, (src, dst, label) in enumerate(trace, start=1):
        label = " ".join(str(label).split())
        label = label if len(label) <= width else label[: width - 1] + "…"
        color = "#c44e52" if "SQLite" in (src, dst) else "#4c72b0"
        ax.annotate("", xy=(x[dst], -i), xytext=(x[src], -i), arrowprops=dict(arrowstyle="-|>", color=color, lw=1.8))
        ax.text((x[src] + x[dst]) / 2, -i + 0.14, f"{i}. {label}", ha="center", va="bottom", fontsize=9,
                bbox=dict(fc="white", ec="none", pad=1.2))
    ax.set_xlim(-0.6, len(actors) - 0.4)
    ax.set_ylim(-len(trace) - 0.5, 1.2)
    ax.axis("off")
    ax.set_title("Tool calling: who talks to whom", fontsize=13, pad=12)
    plt.show()

draw_sequence(trace)
````

Read the diagram: the **LLM never touches SQLite**. It only asks; the App decides whether to obey.

## 10 · Benchmark: latency, tokens and cost
*Choosing a model is an engineering trade-off you can measure*

"Which model is best?" has no answer. "Which model is fast enough, good enough and cheap enough for **this** task?" does — and you find it by measuring. We send the same prompts to both providers and collect latency, token counts and cost into one DataFrame.

Prices change often: check the provider's pricing page and edit `PRICE_PER_MILLION` before trusting the cost column.

````python
# USD per 1 million tokens: (input, output). None = fill in from the provider's pricing page.
PRICE_PER_MILLION = {
    "OpenAI": (0.05, 0.40),     # gpt-5-nano list price, checked Oct 2026
    "NVIDIA": (None, None),     # see the model page on huggingface.co -> Inference Providers
}

PROMPTS = {
    "fact":      "What is the time complexity of binary search? One sentence.",
    "code":      "Write a Python function that checks if a string is a palindrome. Code only.",
    "translate": "Translate to Portuguese: 'The model returned a response in two seconds.'",
    "reason":    "A bat and a ball cost 1.10 in total. The bat costs 1.00 more than the ball. How much is the ball? Answer with the number only.",
}

def call_openai(prompt):
    r = openai_client.responses.create(model=OPENAI_MODEL, input=prompt, reasoning={"effort": "minimal"})
    return r.output_text, r.usage.input_tokens, r.usage.output_tokens

def call_nvidia(prompt):
    text, r = nvidia_chat([{"role": "user", "content": prompt}])
    return text, r.usage.prompt_tokens, r.usage.completion_tokens

rows = []
for provider, call in [("OpenAI", call_openai), ("NVIDIA", call_nvidia)]:
    for task, prompt in PROMPTS.items():
        t0 = time.perf_counter()
        try:
            text, tokens_in, tokens_out = call(prompt)
            seconds = time.perf_counter() - t0
            price_in, price_out = PRICE_PER_MILLION[provider]
            cost = (tokens_in * price_in + tokens_out * price_out) / 1e6 if price_in is not None else float("nan")
            rows.append({"provider": provider, "task": task, "seconds": seconds, "tokens_in": tokens_in,
                         "tokens_out": tokens_out, "tokens_per_s": tokens_out / seconds, "usd": cost,
                         "answer": " ".join(text.split())[:70]})
        except Exception as e:
            print(f"[Erro no provedor {provider} para a tarefa {task}]: {e}")
            rows.append({"provider": provider, "task": task, "seconds": float("nan"), "tokens_in": 0,
                         "tokens_out": 0, "tokens_per_s": 0, "usd": float("nan"),
                         "answer": "Erro de Conexão"})

bench_df = pd.DataFrame(rows)
bench_df.round({"seconds": 2, "tokens_per_s": 1, "usd": 6})
````

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.barplot(data=bench_df, x="task", y="seconds", hue="provider", ax=axes[0])
axes[0].set(title="Latency per task", xlabel="task", ylabel="seconds (lower is better)")
sns.barplot(data=bench_df, x="task", y="tokens_per_s", hue="provider", ax=axes[1])
axes[1].set(title="Generation speed per task", xlabel="task", ylabel="output tokens per second (higher is better)")
plt.tight_layout()
plt.show()

bench_df.groupby("provider")[["seconds", "tokens_out", "tokens_per_s", "usd"]].agg(
    {"seconds": "mean", "tokens_out": "sum", "tokens_per_s": "mean", "usd": lambda s: s.sum(min_count=1)}).round(4)
````

## Wrap-up
*What to remember when the notebook is closed*

1. An LLM API is **HTTPS + JSON**: endpoint, bearer key, request body, response body.
2. The protocol is (almost) a standard: **changing provider = changing `base_url`, key and model name**.
3. The API is **stateless**; memory is the history you resend, and you pay for it in tokens.
4. **Streaming** changes perceived speed, **temperature** changes variety, neither changes what the model knows.
5. **Structured outputs** turn text into data; **tool calling** turns a text generator into something that can act — through *your* code.
6. Model choice is an **engineering trade-off**: measure latency, quality and cost on your own task.

````python
def obter_clima(cidade):
    #consulta o clima usando a API wttr.in
    url = f"https://wttr.in/{cidade}?format=j1"

    resposta = requests.get(url)
    if resposta.status_code == 200:
        dados = resposta.json()
        temp = dados["current_condition"][0]["temp_C"]
        return json.dumps({"cidade": cidade, "temperatura": f"{temp}°C"})
    return json.dumps({"erro": "Não foi possível obter o clima."})
````

````python
def obter_cotacao(moeda_origem, moeda_destino):
    #consulta a cotação usando a AwesomeAPI
    url = f"https://economia.awesomeapi.com.br/json/last/{moeda_origem}-{moeda_destino}"

    resposta = requests.get(url)
    if resposta.status_code == 200:
        dados = resposta.json()
        chave = f"{moeda_origem}{moeda_destino}"
        taxa = dados[chave]["bid"]
        return json.dumps({"moeda_origem": moeda_origem, "moeda_destino": moeda_destino, "taxa": taxa})
    return json.dumps({"erro": "Não foi possível obter a cotação."})
````

````python
# schema das ferramentas para o llm
tools = [
    {
        "type": "function",
        "function": {
            "name": "obter_clima",
            "description": "Obtém o clima e temperatura de uma cidade.",
            "parameters": {
                "type": "object",
                "properties": {
                    "cidade": {"type": "string", "description": "Nome da cidade, ex: Londres"}
                },
                "required": ["cidade"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "obter_cotacao",
            "description": "Obtém a taxa de câmbio entre duas moedas em formato ISO 4217 (ex: USD, BRL).",
            "parameters": {
                "type": "object",
                "properties": {
                    "moeda_origem": {"type": "string", "description": "Moeda de origem, ex: BRL"},
                    "moeda_destino": {"type": "string", "description": "Moeda de destino, ex: EUR"}
                },
                "required": ["moeda_origem", "moeda_destino"]
            }
        }
    }
]
````

````python
def agente_local(mensagem_usuario):
    mensagens = [
        {"role": "system", "content": "Você é um agente de viagens no Brasil. Use as ferramentas disponíveis para ajudar o utilizador."},
        {"role": "user", "content": mensagem_usuario}
    ]

    #chama o modelo descarregado localmente com ollama
    resposta_llm = client.chat.completions.create(
        model="llama3.2:3b",
        messages=mensagens,
        tools=tools,
        tool_choice="auto"
    )

    mensagem_assistente = resposta_llm.choices[0].message
    mensagens.append(mensagem_assistente)

    if mensagem_assistente.tool_calls:
        funcoes_disponiveis = {
            "obter_clima": obter_clima,
            "obter_cotacao": obter_cotacao
        }

        for tool_call in mensagem_assistente.tool_calls:
            nome_funcao = tool_call.function.name
            funcao_python = funcoes_disponiveis[nome_funcao]
            argumentos = json.loads(tool_call.function.arguments)

            resultado_api = funcao_python(**argumentos)

            mensagens.append({
                "tool_call_id": tool_call.id,
                "role": "tool",
                "name": nome_funcao,
                "content": resultado_api,
            })

        resposta_final = client.chat.completions.create(
            model="llama3.2:3b",
            messages=mensagens
        )
        return resposta_final.choices[0].message.content

    return mensagem_assistente.content
````
